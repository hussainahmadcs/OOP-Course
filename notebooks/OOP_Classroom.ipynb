## 1. What is object-oriented programming?
**Object-oriented programming (OOP)** organizes code around objects that combine data and behavior.

| Term | Meaning | Example |
|---|---|---|
| Class | A definition used to create objects | `Student` |
| Object (instance) | One particular item created from a class | `student1` |
| Attribute | Data stored in an object | `student1.name` |
| Method | A function defined in a class | `student1.introduce()` |
| State | The current values of an object's attributes | Name = Sara, major = CS |

One `Student` class can create many student objects. Each object can store different data.

## 2. Define a class
`__init__` is the initializer: Python calls it automatically when a new instance is created.  
`self` refers to the instance receiving the method call.  
`self.name = name` stores the parameter `name` as an attribute of that instance.

Indentation places the methods inside the class. Class names conventionally use `PascalCase`.

```python
class Student:
    def __init__(self, name, major):
        self.name = name
        self.major = major

    def introduce(self):
        return f"Hello, I am {self.name}. I study {self.major}."
```

## 3. Create and use objects
`Student(...)` creates an instance. Dot notation accesses its attributes and methods. In `student1.introduce()`, Python supplies `student1` as `self`; you do not pass it yourself.

```python
student1 = Student("Sara", "Computer Science")
student2 = Student("Noor", "Cybersecurity")
print(student1.name)
print(student1.introduce())
print(student2.introduce())
```

## 4. Default attribute values and methods
A default parameter is used when the caller omits that argument. Here, every new counter starts at `0` unless another value is supplied.

`click()` changes state. `display()` returns text. A method without an explicit `return` returns `None`.

```python
class Counter:
    def __init__(self, label, value=0):
        self.label = label
        self.value = value

    def click(self):
        self.value += 1

    def reset(self):
        self.value = 0

    def display(self):
        return f"{self.label}: {self.value}"

visitors = Counter("Visitors")
points = Counter("Points", 10)
visitors.click()
visitors.click()
print(visitors.display())
print(points.display())
visitors.reset()
print(visitors.display())
```

## 5. Worked example — Bank account
An account has an `owner` and a `balance`. Its methods deposit or withdraw money.

- The balance defaults to `0`.
- A deposit must be positive.
- A withdrawal must be positive and no larger than the available balance.
- `True` means the operation succeeded; `False` means it was rejected.

This classroom example uses whole-number amounts and assumes numeric inputs. The public attributes can still be changed directly; these checks apply when using the methods.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        if balance < 0:
            raise ValueError("Starting balance cannot be negative.")
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            return False
        self.balance += amount
        return True

    def withdraw(self, amount):
        if amount <= 0 or amount > self.balance:
            return False
        self.balance -= amount
        return True

    def summary(self):
        return f"{self.owner}: SAR {self.balance}"
```

**Predict the final balances before running:** Sara starts with 100, deposits 50, withdraws 30, then tries to withdraw 200. Noor starts with the default balance.

```python
sara_account = BankAccount("Sara", 100)
noor_account = BankAccount("Noor")
print("Deposit:", sara_account.deposit(50))
print("Withdrawal:", sara_account.withdraw(30))
print("Large withdrawal:", sara_account.withdraw(200))
print(sara_account.summary())
print(noor_account.summary())
```

**Check your understanding:** Sara has SAR 120; Noor has SAR 0. The rejected withdrawal leaves the balance unchanged.

## 6. Reference puzzle — two names, one object
Assigning an object to another variable does **not** copy it. Both variables can refer to the same object. Creating a new object gives separate state.

```python
wallet = BankAccount("Hana", 100)
alias = wallet
alias.deposit(20)
print(wallet.balance)       # 120: both names refer to the same object
print(alias is wallet)     # True: identity comparison

alias = BankAccount("Hana", 0)
alias.deposit(5)
print(wallet.balance)      # 120: the original account still exists
print(alias.balance)       # 5: this is a different account
print(alias is wallet)     # False
```
